# 04. Universe Filtering with Metadata
Demonstrates how to use `symbol_information.parquet` to filter the universe before loading the panel.

```python
import sys
sys.path.append('..')
from loader import load_panel, load_metadata
import polars as pl
from datetime import datetime

# 1. Load Metadata
meta = load_metadata()

# 2. Filter Universe
# Example: Select symbols listed before 2022-06-01
cutoff_date = datetime(2022, 6, 1)
valid_symbols = (
    meta
    .filter(pl.col("listing_time") < cutoff_date)
    .select("symbol")
    .to_series()
    .to_list()
)
print(f"Selected {len(valid_symbols)} symbols listed before {cutoff_date.date()}")

# 3. Load Panel with Filtered Universe
if valid_symbols:
    lf = load_panel(start_date="2022-01-01", end_date="2022-02-01", symbols=valid_symbols)
    df = lf.collect()
    print("\nLoaded Panel:")
    print(df.head())
else:
    print("No symbols matched criteria.")
```
