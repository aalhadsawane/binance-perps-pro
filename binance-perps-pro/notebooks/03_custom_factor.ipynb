# 03. Custom Factor Example
Template for implementing your own alpha factors.

```python
import sys
sys.path.append('..')
from loader import load_panel
import polars as pl

# Load full history
lf = load_panel()

# Custom Factor: Volume-Weighted Return (VWAP proxy momentum)
# Calculate hourly VWAP and compare to Close
factor = (
    lf
    .with_columns([
        (pl.col("dollar_volume") / pl.col("volume")).alias("vwap_1h")
    ])
    .with_columns([
        (pl.col("close") / pl.col("vwap_1h") - 1).alias("vwap_deviation")
    ])
    .select(["timestamp", "symbol", "vwap_deviation"])
    .collect()
)

print(factor.describe())
```
